**- Datos Repetidos**

Es casi imposible que existan 2 viviendas con exactamente las mismas características y vendidas al mismo precio, ya que serían viviendas idénticas, por ello se ha decidido que si hay 2 registros con los mismos valores en todas las variables se tratarán como "duplicadas" y se eliminará uno de los registros repetidos del dataset quedándonos solamente con registros únicos.


```python
# Para eliminar duplicados, las características principales tienen que ser todos los atributos del dataset.
# Se utilizará la función "drop_duplicates" que permite indicar los campos en los que se tienen que encontrar los valores repetidos simultánemanete. 
caracteristicas_pples = datos.columns.values

print("Las características principales son:", caracteristicas_pples, "\n")

data_sin_reps = datos.drop_duplicates(subset=caracteristicas_pples)

print("Se han encontrado",datos.shape[0]-data_sin_reps.shape[0], "registros repetidos que han sido eliminados quedándonos con un total de",data_sin_reps.shape[0], "registros. \n")
```
